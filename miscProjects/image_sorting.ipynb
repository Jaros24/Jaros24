```python
import os
import exifread
```

```python
pic_dir = 'C:/Users/Adam/OneDrive - Tenundra, Inc/Pictures/Camera Roll/2017/'
year = '2017'
```

```python
def get_date(file:str) -> str:
    dates = []
    with open(file, 'rb') as f:
        tags = exifread.process_file(f)
        for tag in tags.keys():
            if 'DateTime' in tag:
                dates.append(str(tags[tag]))
    
    if len(dates) == 0:
        return '' # if no date is found, return an empty string
    
    # should be formatted as 'YYYY:MM:DD HH:MM:SS'
    year = [d[:4] for d in dates]
    month = [d[5:7] for d in dates]
    day = [d[8:10] for d in dates]
    
    # if all the dates are the same, return the first one
    if year.count(year[0]) == len(year) and month.count(month[0]) == len(month) and day.count(day[0]) == len(day):
        return f'{year[0]}{month[0]}{day[0]}' # YYYYMMDD
    else:
        return '' # if the dates are different, return an empty string
```

```python
for filename in os.listdir(pic_dir):
    date = get_date(pic_dir + filename)
    
    if date != '':
        if date in filename:
            continue
        else:
            os.rename(pic_dir + filename, pic_dir + f'{date}_{filename}')
```
